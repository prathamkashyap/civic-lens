```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

fig, ax = plt.subplots(figsize=(10, 14))
ax.set_xlim(0, 10)
ax.set_ylim(0, 15)
ax.axis('off')

# Define stages with positions and text
stages = [
    (5, 13.5, "Stage 1: Complaint Registration", "Citizen submits complaint via web/mobile form"),
    (5, 11.5, "Stage 2: Validation & Storage", "Data validated and stored in Firebase database"),
    (5, 9.5, "Stage 3: Feature Computation", "Calculate spatial density, temporal features, text embeddings"),
    (5, 7.5, "Stage 4: ML Priority Prediction", "Random Forest classifies: HIGH / MEDIUM / LOW"),
    (5, 5.5, "Stage 5: Hotspot Detection", "DBSCAN identifies geographic complaint clusters"),
    (5, 3.5, "Stage 6: Visualization", "Display heatmaps, charts, priority lists on dashboard"),
    (5, 1.5, "Stage 7: Resource Allocation", "Municipal staff prioritize and assign interventions")
]

# Draw boxes and text
for x, y, title, desc in stages:
    # Draw box
    box = FancyBboxPatch((x-4.5, y-0.7), 9, 1.4, 
                          boxstyle="round,pad=0.15", 
                          edgecolor='#1f77b4', 
                          facecolor='#e3f2fd',
                          linewidth=2.5)
    ax.add_patch(box)
    
    # Add title (bold)
    ax.text(x, y+0.25, title, ha='center', va='center', 
            fontsize=11, weight='bold', color='#0d47a1')
    
    # Add description
    ax.text(x, y-0.25, desc, ha='center', va='center', 
            fontsize=9, color='#424242', style='italic')
    
    # Draw arrow to next stage (except last)
    if y > 1.5:
        arrow = FancyArrowPatch((x, y-0.75), (x, y-1.25),
                               arrowstyle='->', mutation_scale=25, 
                               color='#1f77b4', linewidth=3)
        ax.add_patch(arrow)

# Title
plt.title("Working Principle: Smart Civic Issue Register & Tracker", 
          fontsize=16, weight='bold', pad=30, color='#0d47a1')

# Add figure number
ax.text(5, 0.3, "Figure 4.1: System Workflow Overview", 
        ha='center', fontsize=10, style='italic', color='#666')

plt.tight_layout()
plt.savefig('workflow_diagram.png', dpi=300, bbox_inches='tight', facecolor='white')
print("✅ Workflow diagram saved as 'workflow_diagram.png'")
plt.show()
```
